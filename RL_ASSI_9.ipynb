```python
# Assignment 1: Algebraic Solution of Bellman Expectation Equations

gamma = 0.9

# Solve the simultaneous equations:
# V(S0) = -1 + 0.81 * V(S1)
# V(S1) = 89.9 + 0.09 * V(S0)

V_S0 = 71.819 / 0.9271
V_S1 = 89.9 + 0.09 * V_S0

# Calculate action values
Q_S0_Down = -1 + gamma * V_S1
Q_S0_Right = -1

Q_S1_Right = 100
Q_S1_Up = -1 + gamma * V_S0

# Verify state values using the policy
V_S0_check = 0.9 * Q_S0_Down + 0.1 * Q_S0_Right
V_S1_check = 0.9 * Q_S1_Right + 0.1 * Q_S1_Up

# Display results
print("ASSIGNMENT 1: ALGEBRAIC SOLUTION")
print("--------------------------------")
print(f"V(S0) = {V_S0:.2f}")
print(f"V(S1) = {V_S1:.2f}")
print(f"Q(S0, Down) = {Q_S0_Down:.2f}")
print(f"Q(S0, Right) = {Q_S0_Right:.2f}")
print(f"Q(S1, Right) = {Q_S1_Right:.2f}")
print(f"Q(S1, Up) = {Q_S1_Up:.2f}")

print("\nVerification:")
print(f"V(S0) from action values = {V_S0_check:.2f}")
print(f"V(S1) from action values = {V_S1_check:.2f}")
```

```python
# Assignment 2: Iterative Policy Evaluation

gamma = 0.9
theta = 0.000001

# Initialize state values
V_S0 = 0.0
V_S1 = 0.0

iterations = 0

# Iterative policy evaluation
while True:
    # Use previous state values for both updates
    new_V_S0 = (
        0.1 * (-1) +
        0.9 * (-1 + gamma * V_S1)
    )

    new_V_S1 = (
        0.9 * 100 +
        0.1 * (-1 + gamma * V_S0)
    )

    # Calculate maximum change
    delta = max(
        abs(new_V_S0 - V_S0),
        abs(new_V_S1 - V_S1)
    )

    # Update state values
    V_S0 = new_V_S0
    V_S1 = new_V_S1

    iterations += 1

    # Stop when values converge
    if delta < theta:
        break

# Calculate action values after convergence

# S0, Down: Wall (0.1), S1 (0.9)
Q_S0_Down = (
    0.1 * (-1) +
    0.9 * (-1 + gamma * V_S1)
)

# S0, Right: Wall (1.0)
Q_S0_Right = -1

# S1, Right: Goal (0.9), S0 (0.1)
Q_S1_Right = (
    0.9 * 100 +
    0.1 * (-1 + gamma * V_S0)
)

# S1, Up: S0 (1.0)
Q_S1_Up = -1 + gamma * V_S0

# Display results
print("ASSIGNMENT 2: ITERATIVE POLICY EVALUATION")
print("------------------------------------------")
print("Number of iterations:", iterations)

print(f"V(S0) = {V_S0:.5f}")
print(f"V(S1) = {V_S1:.5f}")

print("\nAction Values:")
print(f"Q(S0, Down) = {Q_S0_Down:.5f}")
print(f"Q(S0, Right) = {Q_S0_Right:.5f}")
print(f"Q(S1, Right) = {Q_S1_Right:.5f}")
print(f"Q(S1, Up) = {Q_S1_Up:.5f}")

# Verify policy consistency
print("\nPolicy Consistency Checks:")
print(f"V(S0) = Q(S0, Down): {abs(V_S0 - Q_S0_Down) < theta}")
print(f"V(S1) = Q(S1, Right): {abs(V_S1 - Q_S1_Right) < theta}")
```
