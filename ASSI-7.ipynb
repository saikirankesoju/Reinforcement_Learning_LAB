```python



n = int(input("Enter the number of time steps: "))

# Accept rewards
rewards = []

for i in range(n):
    reward = float(input(f"Enter reward for time step {i + 1}: "))
    rewards.append(reward)

# Calculate undiscounted return
undiscounted_return = sum(rewards)

# Accept discount factor
gamma = float(input("Enter discount factor (gamma): "))

# Calculate discounted return
discounted_return = 0

for i in range(n):
    discounted_return += rewards[i] * (gamma ** i)

# Display results
print("\n--- Results ---")
print("Entered reward values:", rewards)
print("Undiscounted Return:", undiscounted_return)
print("Discounted Return:", discounted_return)3
```

```python
# Assignment 2: 2 x 3 Grid World MDP

# ------------------------------------------------
# 1. Define State Space
# ------------------------------------------------
states = ["S0", "S1", "S3", "S4", "G", "X"]

# ------------------------------------------------
# 2. Define Action Space
# ------------------------------------------------
actions = ["Up", "Down", "Left", "Right"]

# ------------------------------------------------
# 3. Define Reward Function
# ------------------------------------------------
def reward_function(state):
    if state == "G":
        return 10
    elif state == "X":
        return -5
    else:
        return -1

# ------------------------------------------------
# 4. Define Discount Factor
# ------------------------------------------------
gamma = 0.9

# ------------------------------------------------
# 5. Define Transition Function
# ------------------------------------------------
transitions = {
    "S0": {
        "Up": "S0",
        "Down": "S0",
        "Left": "S0",
        "Right": "S1"
    },

    "S1": {
        "Up": "S1",
        "Down": "S3",
        "Left": "S0",
        "Right": "G"
    },

    "S3": {
        "Up": "S1",
        "Down": "S3",
        "Left": "S3",
        "Right": "S4"
    },

    "S4": {
        "Up": "G",
        "Down": "S4",
        "Left": "S3",
        "Right": "S4"
    },

    "G": {
        "Up": "G",
        "Down": "G",
        "Left": "G",
        "Right": "G"
    },

    "X": {
        "Up": "X",
        "Down": "X",
        "Left": "X",
        "Right": "X"
    }
}

# ------------------------------------------------
# 6. Estimate Next State
# ------------------------------------------------
def transition(state, action):
    return transitions[state][action]

# ------------------------------------------------
# Display MDP
# ------------------------------------------------
print("===== 2 x 3 GRID WORLD MDP =====")

print("\nState Space:")
print(states)

print("\nAction Space:")
print(actions)

print("\nDiscount Factor:")
print(gamma)

print("\nReward Values:")
for state in states:
    print(state, "=", reward_function(state))

# ------------------------------------------------
# Display all transitions
# ------------------------------------------------
print("\n===== TRANSITIONS =====")

for state in states:
    for action in actions:
        next_state = transition(state, action)
        print(state, "+", action, "->", next_state)

# ------------------------------------------------
# User Input
# ------------------------------------------------
print("\n===== NEXT STATE ESTIMATION =====")

current_state = input("Enter current state: ")
action = input("Enter action: ")

if current_state in states and action in actions:
    next_state = transition(current_state, action)

    print("\nCurrent State:", current_state)
    print("Action:", action)
    print("Next State:", next_state)
    print("Reward:", reward_function(next_state))
else:
    print("Invalid state or action!")
```
