```python
import random

# Part A
machines = ['A', 'B', 'C']

# Number of plays and exploration probability
N_play = 10
epsilon = 0.2

# Part B
N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': 0, 'B': 0, 'C': 0}
Q = {'A': 0, 'B': 0, 'C': 0}

total_reward = 0

# Part C to Part E – Modified ε-Greedy Strategy
for play in range(1, N_play + 1):

    # Generate random number
    r = random.random()

    # Highest Q-value machine
    highest_Q = max(Q.values())

    # Machines having highest Q-value
    best_machines = [m for m in machines if Q[m] == highest_Q]

    if r < epsilon:
        # Exploration:
        # Select a machine except the highest Q machine
        available = [m for m in machines if m not in best_machines]

        # If all machines are considered highest initially,
        # randomly select any machine
        if len(available) == 0:
            selected = random.choice(machines)
        else:
            selected = random.choice(available)

        decision = "Exploration"

    else:
        # Exploitation:
        selected = random.choice(best_machines)
        decision = "Exploitation"

    # Part D – Reward Collection
    print("\nPlay", play)
    print("Random number:", round(r, 3))
    print("Decision:", decision)
    print("Play with machine", selected)

    reward = float(input("Enter reward: "))

    # Part E – Update N, R and Q
    N[selected] += 1
    R[selected] += reward

    # Incremental sample-average formula
    Q[selected] = Q[selected] + (reward - Q[selected]) / N[selected]

    total_reward += reward

    print("Updated Q-value of", selected, "=", round(Q[selected], 2))


# Part F – Display Results
print("\n======================================")
print("FINAL RESULTS")
print("======================================")

print("\nMachine\tN\tR\tQ")

for machine in machines:
    print(machine, "\t", N[machine], "\t", R[machine],
          "\t", round(Q[machine], 2))

print("\nTotal Reward =", total_reward)
```
