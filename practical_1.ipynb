```python
#Question 1:
def greedy_arithmetic_mean():
    machines = {
        'A': [3, 2, 4],
        'B': [6, 7],
        'C': [5, 4, 6]
    }

    q_values = {}

    for machine, rewards in machines.items():
        q_a = sum(rewards) / len(rewards) if rewards else 0.0
        q_values[machine] = q_a

        print(f"Machine {machine}")
        print(f"Rewards: {rewards}")
        print(f"Q({machine}) = {q_a:.2f}")
        print("-" * 20)

    best_machine = max(q_values, key=q_values.get)

    print(f"Greedy Selected Machine: {best_machine}")

if __name__ == "__main__":
    print("--- QUESTION 1: ARITHMETIC MEAN ---")
    greedy_arithmetic_mean()
```

```python
#Question 2:
def greedy_incremental_update():
    observed_plays = {
        'A': [3, 2, 4],
        'B': [6, 7],
        'C': [5, 4, 6]
    }

    q_values = {machine: 0.0 for machine in observed_plays.keys()}
    n_values = {machine: 0 for machine in observed_plays.keys()}

    print("Initial State:")
    print(f"Q-values: {q_values}")
    print(f"N-values: {n_values}\n")
    print("-" * 30)

    for machine, rewards in observed_plays.items():
        for reward in rewards:

            print(f"Selected Machine: {machine} | Observed Reward: {reward}")

            n_values[machine] += 1

            q_old = q_values[machine]
            n_a = n_values[machine]
            q_new = q_old + (reward - q_old) / n_a

            q_values[machine] = q_new


            current_best = max(q_values, key=q_values.get)

            print(f"Updated N({machine}) = {n_values[machine]}")
            print(f"Updated Q({machine}) = {q_values[machine]:.2f}")
            print(f"Current Highest Q(a) Machine: {current_best}")
            print("-" * 30)

    print("\nFinal Q-values after all plays:")
    for m, q in q_values.items():
        print(f"Q({m}) = {q:.2f}")

if __name__ == "__main__":
    print("--- QUESTION 2: INCREMENTAL UPDATE ---")
    greedy_incremental_update()
```
