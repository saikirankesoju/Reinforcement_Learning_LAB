```python
import math

# Part A – Machine Initialization
machines = ['A', 'B', 'C']

# Number of plays
N_play = 10

# Part B – Initialize N, R, UCB and Q tables
N = {'A': 0, 'B': 0, 'C': 0}       # Number of times played
R = {'A': 0, 'B': 0, 'C': 0}       # Total reward
Q = {'A': 0, 'B': 0, 'C': 0}       # Estimated reward
UCB = {'A': 0, 'B': 0, 'C': 0}     # UCB values

total_reward = 0

# Part C – UCB Calculation and Selection
for play in range(1, N_play + 1):

    # Calculate UCB for each machine
    for machine in machines:

        # If machine has not been played, give infinity
        if N[machine] == 0:
            UCB[machine] = float('inf')
        else:
            UCB[machine] = Q[machine] + math.sqrt(
                (2 * math.log(play)) / N[machine]
            )

    # Select machine with highest UCB
    selected = max(machines, key=lambda x: UCB[x])

    print("\nPlay", play)
    print("UCB Values:")

    for machine in machines:
        if UCB[machine] == float('inf'):
            print(machine, "= Infinity")
        else:
            print(machine, "=", round(UCB[machine], 3))

    print("Play with machine", selected)

    # Part D – Reward Collection
    reward = float(input("Enter reward: "))

    # Part E – Update N, R and Q
    N[selected] += 1
    R[selected] += reward

    # Incremental sample-average formula
    Q[selected] = Q[selected] + (
        reward - Q[selected]
    ) / N[selected]

    total_reward += reward

    print("Updated Q-value =", round(Q[selected], 3))


# Part F – Display Final Results
print("\n==========================================")
print("FINAL RESULTS")
print("==========================================")

print("\nMachine\tN\tR\tUCB\tQ")

# Calculate final UCB values
for machine in machines:
    if N[machine] == 0:
        UCB[machine] = float('inf')
    else:
        UCB[machine] = Q[machine] + math.sqrt(
            (2 * math.log(N_play)) / N[machine]
        )

for machine in machines:
    if UCB[machine] == float('inf'):
        ucb_value = "Infinity"
    else:
        ucb_value = round(UCB[machine], 3)

    print(
        machine,
        "\t", N[machine],
        "\t", R[machine],
        "\t", ucb_value,
        "\t", round(Q[machine], 3)
    )

print("\nTotal Reward =", total_reward)

# Analysis
best_machine = max(machines, key=lambda x: Q[x])

print("\nAnalysis:")
print("Best machine based on Q-value:", best_machine)
print("Highest estimated reward:", round(Q[best_machine], 3))
```
