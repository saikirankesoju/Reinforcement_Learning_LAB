```python
import random

machines = ['A', 'B', 'C']
N_play = 10
epsilon = 0.2
N = [0, 0, 0]
R = [0, 0, 0]
Q = [0.0, 0.0, 0.0]
total_reward = 0

# Define reward distributions for each machine for simulation purposes
reward_distributions = {
    'A': {'mean': 3.0, 'std': 1.0},
    'B': {'mean': 2.0, 'std': 0.5},
    'C': {'mean': 5.0, 'std': 1.5}
}

for play in range(N_play):
    r_epsilon = random.random()
    if r_epsilon < epsilon:
        index = random.randint(0, 2)
        print("Exploration")
    else:
        # In case all Q values are the same (e.g., all 0.0 initially), random.choice helps
        # prevent always picking the first one if there are multiple max values.
        max_q_value = max(Q)
        best_machines_indices = [i for i, q_val in enumerate(Q) if q_val == max_q_value]
        index = random.choice(best_machines_indices)
        print("Exploitation")

    machine = machines[index]

    # Simulate reward instead of asking for input
    # Generate a reward from a normal distribution based on the machine's defined distribution
    mean = reward_distributions[machine]['mean']
    std = reward_distributions[machine]['std']
    reward = random.gauss(mean, std)
    # Ensure reward is non-negative for this simulation
    reward = max(0, reward)

    print("\nPlay", play + 1)
    print("Machine selected:", machine)
    print(f"Simulated reward: {reward:.2f}") # Display the simulated reward

    N[index] = N[index] + 1
    R[index] = R[index] + reward
    Q[index] = Q[index] + (reward - Q[index]) / N[index]
    total_reward = total_reward + reward

    print("N =", N)
    print("R =", [round(val, 2) for val in R]) # Round R for better display
    print("Q =", [round(val, 2) for val in Q]) # Round Q for better display

print("\n========== FINAL RESULT ==========")
for i in range(3):
    print(
        "Machine", machines[i],
        ": N =", N[i],
        ", R =", round(R[i], 2),
        ", Q =", round(Q[i], 2)
    )
print("\nTotal Reward after", N_play, "plays =", round(total_reward, 2))
```
